```python
import numpy as np
import pandas as pd
import sklearn

print(f"numpy: {np.__version__} | pandas: {pd.__version__} | sklearn: {sklearn.__version__}")

url = "https://raw.githubusercontent.com/awais-DS/Data/refs/heads/main/sales_data.csv"
df = pd.read_csv(url)

print(f"Dataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
display(df.head())

overview = pd.DataFrame({
    'Dtype': df.dtypes,
    'Non_Null': df.notnull().sum(),
    'Missing': df.isnull().sum(),
    'Missing_%': (df.isnull().sum() / len(df) * 100).round(2),
    'Unique': df.nunique()
})

print(f"Exact Duplicate Rows: {df.duplicated().sum()}")
display(overview)

print("--- Numerical Summary ---")
display(df.describe().T)

print("--- Categorical Summary ---")
display(df.describe(include=['object', 'string']).T)

df['Date'] = pd.to_datetime(df['Date'])

print(f"Date Range:          {df['Date'].min().date()} to {df['Date'].max().date()}")
print(f"Total Stores:        {df['Store ID'].nunique()} ({sorted(df['Store ID'].unique())})")
print(f"Total Products:      {df['Product ID'].nunique()}")
print(f"Categories:          {list(df['Category'].unique())}")
print(f"Regions:             {list(df['Region'].unique())}")
print("-" * 45)
print(f"Negative Prices:     {(df['Price'] < 0).sum()}")
print(f"Negative Demand:     {(df['Demand'] < 0).sum()}")
print(f"Sales > Inventory:   {(df['Units Sold'] > df['Inventory Level']).sum()}")
print(f"Stockout Events:     {(df['Demand'] > df['Inventory Level']).sum()} ({(df['Demand'] > df['Inventory Level']).mean()*100:.1f}% of rows)")
```
